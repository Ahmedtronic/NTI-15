```python
import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
text = "The quick brown fox jumps over the lazy dog!."

doc = nlp(text)

for token in doc:
    print(f"{token.text:<12} POS: {token.pos_:<10} | Detailed Tag: {token.tag_} | Explanation: {spacy.explain(token.tag_)}")
```

```python
text = "Barack Obama was born in Hawaii and served as president of the United States and took 5 dollars from me in the World Cup."
doc = nlp(text)


for ent in doc.ents:
    print(f"{ent.text} ({ent.label_})")
```

```python
import nltk
from nltk import word_tokenize, pos_tag, ne_chunk

# Ensure the required NLTK data is downloaded
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('maxent_ne_chunker_tab') # Added this line to download the missing resource

# Sample text
text = "Barack Obama was born in Hawaii and served as president of the United States."

# Tokenize and POS tagging
tokens = word_tokenize(text)
pos_tags = pos_tag(tokens)

# Named Entity Recognition
tree = ne_chunk(pos_tags)

# Extract named entities
print("Named Entities:")
for subtree in tree:
    if hasattr(subtree, 'label'):
        entity = " ".join([token for token, pos in subtree])
        print(f"{entity} ({subtree.label()})")
```

```python
from sumy.parsers.plaintext import PlaintextParser
from sumy.nlp.tokenizers import Tokenizer
from sumy.summarizers.lsa import LsaSummarizer  # You can try Luhn, LexRank, etc.

text = """
Artificial Intelligence is transforming industries, from healthcare to finance.
It enables machines to learn from data, identify patterns, and make decisions with minimal human intervention.
AI technologies such as natural language processing, computer vision, and robotics are increasingly integrated into our daily lives.
"""

parser = PlaintextParser.from_string(text, Tokenizer("english"))
summarizer = LsaSummarizer()

summary = summarizer(parser.document, 2)  # 2 = number of sentences in summary

print("Summary:")
for sentence in summary:
    print("-", sentence)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

texts = [
    "The cat sat on the mat.",
    "A cat is sitting on the rug."
]

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(texts)

print(tfidf_matrix[0:1])
print(tfidf_matrix[1:2])

# Calculate cosine similarity

similarity = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[1:2])
print("Cosine Similarity:", similarity[0][0])
```

```python
from transformers import pipeline

# Load the QA pipeline with a pretrained model
qa_pipeline = pipeline("question-answering", model="distilbert-base-cased-distilled-squad")

# Your context paragraph
context = """
The Nile is a major north-flowing river in northeastern Africa.
It is commonly regarded as the longest river in the world.
It flows through eleven countries, including Uganda, Sudan, and Egypt.
"""


question = "Which countries does the Nile flow through?"


result = qa_pipeline(
    context=context,
    question=question
)

print("Answer:", result['answer'])
```

```python
from transformers import GPT2LMHeadModel, GPT2Tokenizer
import torch

# Load pre-trained model and tokenizer
model_name = "gpt2"
tokenizer = GPT2Tokenizer.from_pretrained(model_name)
model = GPT2LMHeadModel.from_pretrained(model_name)

# Encode the prompt
input_ids = tokenizer.encode("In a future where humans and AI work together,", return_tensors='pt')

# Generate text
output = model.generate(input_ids, max_length=100, num_return_sequences=1, temperature=0.7)

# Decode and print
print(tokenizer.decode(output[0], skip_special_tokens=True))
```

```python
from transformers import AutoTokenizer, AutoModelForQuestionAnswering
import torch

model_name = "distilbert-base-cased-distilled-squad"

# Load tokenizer and model
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForQuestionAnswering.from_pretrained(model_name)

context = """
The Nile is a major north-flowing river in northeastern Africa.
It is commonly regarded as the longest river in the world.
It flows through eleven countries, including Uganda, Sudan, and Egypt.
"""

question = "Which country does the Nile flow through?"

# Tokenize
inputs = tokenizer(
    question,
    context,
    return_tensors="pt"
)

# Run model
with torch.no_grad():
    outputs = model(**inputs)

# Get answer span
start = torch.argmax(outputs.start_logits)
end = torch.argmax(outputs.end_logits)

# Convert tokens back to text
answer = tokenizer.decode(
    inputs["input_ids"][0][start:end + 1],
    skip_special_tokens=True
)

print("Answer:", answer)
```
