```python
import pandas as pd
from openpyxl import Workbook

# Reproducible fake sales data
data = {
    "Product_ID": [f"P{i:03d}" for i in range(1, 11)],
    "Product_Name": [
        "Laptop", "Monitor", "Keyboard", "Mouse", "Headset",
        "Webcam", "Printer", "SSD", "RAM", "Router"
    ],
    "Category": [
        "Computers", "Computers", "Accessories", "Accessories", "Accessories",
        "Accessories", "Printers", "Storage", "Components", "Networking"
    ],
    "Unit_Price": [35000, 12000, 1800, 900, 2500, 3200, 8500, 4500, 2800, 2200],
}

# Slightly different quantities for each month
month_quantities = {
    "month_1_sales": [12, 18, 35, 48, 22, 16, 9, 27, 31, 24],
    "month_2_sales": [15, 21, 42, 53, 25, 19, 11, 32, 36, 29],
    "month_3_sales": [18, 24, 47, 61, 29, 22, 13, 38, 41, 34],
}

with pd.ExcelWriter("quarter_sales_fake_data.xlsx", engine="openpyxl") as writer:
    for sheet_name, quantities in month_quantities.items():
        df = pd.DataFrame(data)
        df["Quantity_Sold"] = quantities
        df["Total_Sales"] = df["Unit_Price"] * df["Quantity_Sold"]
        df.to_excel(writer, sheet_name=sheet_name, index=False)

print("Created: /mnt/data/quarter_sales_fake_data.xlsx")
print("Sheets: month_1_sales, month_2_sales, month_3_sales")
```
