```python
def Read_Database( database_name ) : 
    conn = pymssql.connect(
    server='localhost',
    port=1433,
    user='sa',
    password='YourStrong!Passw0rd',
    database= database_name )

    cursor = conn.cursor()

    tables_query = """SELECT TABLE_NAME 
    FROM INFORMATION_SCHEMA.TABLES 
    WHERE TABLE_TYPE = 'BASE TABLE';""" 

    cursor.execute(tables_query)

    tables = [ ]
    for row in cursor.fetchall():
        tables.append(row[0])

    database = dict()

    for table in tables: 
        query = f"SELECT * from {table}"
        database[table] = pd.read_sql(query, conn)

    conn.close()
    return database 
```

```python
DEPI_Database = Read_Database( "DEPI" )
```
